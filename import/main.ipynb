```python
from graphics import rectangle,circle
from graphics.graphics_3d import cuboid,sphere
while(1):
    print("1.rectangle\n 2.circle\n 3.cuboid\n 4.sphere\n 5.exit\n")
    ch=int(input("enter your choice:"))
    if(ch==1):
        l=int(input("enter  length"))
        b=int(input("enter breadth"))
        print("area=",rectangle.rectarea(l,b))
        print("perimeter=",rectangle.rectperi(l,b))
    elif(ch==2):
        r=int(input("enter radius:"))
        print("area=",circle.circlearea(r))
        print("perimeter=",circle.circleperimeter(r))
    elif(ch==3):
        l=int(input("enter length:"))
        b=int(input("enter breadth:"))
        h=int(input("enter height:"))
        print("surface area=",cuboid.cuboidarea(l,b,h))
        print("perimeter=",cuboid.perimeter(l,b,h))
    elif(ch==4):
        r=int(input("enter radius of sphere:"))
        print("area=",sphere.spherearea(r))
        print("perimeter=",sphere.sphereperimeter(r))
    elif(ch==5):
        break
    else:
        print("invalid choice")
    
```
