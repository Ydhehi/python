```python
names=input("enter first names:")
count=0
for name in names:
    count=count+name.lower().count('a')
print("no of occurences of a:",count)
```

```python
color_list1=['red','green','blue','lavender']
color_list2=['green','yellow','blue']
for color in color_list1:
    if color not in color_list2:
        print(color)
```

```python
d={'b':15,'a':10,'d':40,'c':40}
print("Ascending order:",dict(sorted(d.items())))
print("Descending order:",dict(sorted(d.items(),reverse=True)))
```
