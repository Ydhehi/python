```python
a=list(map(int,input("enter first list:").split()))
b=list(map(int,input("enter second list:").split()))
if len(a)==len(b):
    print("same length")
else:
    print("different length")
if sum(a)==sum(b):
    print("same sum")
else:
    print("different sum")
for i in a:
    if i in b:
        print("common element",i)
```
